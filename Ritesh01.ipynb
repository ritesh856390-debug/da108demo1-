```python
import matplotlib.pyplot as plt

# Define data for the plot
x = [1, 2, 3, 4, 5] #x-value
y = [2, 3, 5, 7, 11] # y-value

# Create the line plot
plt.plot(x, y, marker='o', linestyle='-', color='g', label='Data')

#Add labels to the axes
plt.xlabel('X-axis') # Add a label to the x-axis
plt.ylabel('Y-axis') #Add a label to the y-axis


#ADD a title to the plot
plt.title('A Line Plot')

#Add a legend
plt.legend()

#Show the plot
plt.show()
```

oop style

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2, 100) #sample data

#OO-Style# we use `.pyplot.figure` to create the figure.
fig, ax= plt.subplots(figsize=(5, 4))
ax.plot(x, x, label='linear') # plot some data on the axes.
ax.plot(x, x**2, label='quadratic') # plot more data on the axes.
ax.plot(x, x**3, label='cubic') # and some more.
ax.set_xlabel('x label') # Add an x-label to the axes.
ax.set_ylabel('y label') # Add a y-label to the axes.
ax.set_title("Simple Plot") # Add a title to the axes.
ax.legend() # Add a legend.
plt.show()
```

Histogram chart

```python
import matplotlib.pyplot as plt
import numpy as np

# Genrate random value
data = np.random.randn(1000)

# to create the histogram
plt.hist(data, bins=30)

#Add labels and title
plt.xlabel('value')
plt.ylabel('Frequency')
plt.title('HISTOGRAM')

#Display the plot
plt.show()
```

pie chart

```python
import matplotlib.pyplot as plt
#DATA
labels = ['a', 'b', 'c', 'd',]
sizes = [20, 30, 40, 50]

# Create a  pie chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%')

# ADD TITLE
plt.title('pie chart')
plt.show()
```

BOX PLOT

```python
import matplotlib.pyplot as plt
import numpy as np

#Genrate some random data
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

# Create a box plot
plt.boxplot(data)

#Add labels and title
plt.xlabel('Groups')
plt.ylabel('Values')
plt.title('Box plot')
plt.show()
```

```python
import time

def release_announcement(message, delay=0.05):
    for char in message:
        print(char, end='', flush=False)
        time.sleep(delay)
    print('\n')

message = "Let's dive in to the DA-108 final week content!"
print("Initializing release announcement sequence...")
time.sleep(1)
print("3")
time.sleep(1)
print("2")
time.sleep(1)
print("1")
time.sleep(1)

release_announcement(message)
print("Release announcement complete! Happy learning!")
```
